# Students Management System

#### Add a student

```python
students = []
def addStudent(student):
    students.append(student)
    print("added successfully")
```

#### Delete a student

```python
def delStudent(student):
    if student in students:
        students.remove(student)
        print("deleted successfully")
    else:
        print("Student not found.")
```

```python
def searchStudent(student):
    if student in students:
        print(f"{student} found on index : {students.index(student)} ")
    else:
        print(f"{student} not found.")
```

```python
print("""
Welcome to SMS!
 Write student's name with add, remove, or search keyword to perform action 
""")
student = input("Enter a student name: ").lower()
if "add" in student:
    addStudent(student.replace("add", "").strip())
elif "remove" in student or "delete" in student:
    delStudent(student.replace("remove", "").replace("delete", "").strip())
elif "search" in student:
    searchStudent(student.replace("search", "").strip())
else:
    print("Invalid action. Please include add, remove, delete, or search.")
```
